# Data Frame

```python
import pandas as pd
```

```python
l=[1,2,3,4,5]
var=pd.DataFrame(l)
print(var)
```

`We can also use Dictionary for Data_Frame`

```python
d={"Name":['Atif','Ahmad','Umer','Aslam'],
   "age":[20,21,19,22]
}
var2=pd.DataFrame(d)
var2
```

# Call a specific Column

```python
var2=pd.DataFrame(d,columns=["Name"])
print(var2)
```

```python
var2=pd.DataFrame(d,columns=["age"])
var2
```

`Call multiple columns`

```python
var2=pd.DataFrame(d,columns=['age','Name'])
print(var2)
```

```python
print(var2["Name"][0])
```
