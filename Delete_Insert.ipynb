# Delete and Insert Function

# `Insert`

- we can insert dictionary items using the `insert` method

```python
dict = {}
# Example of inserting items into a dictionary
def insert(key, value):
    dict[key] = value
insert(1, "List")
insert(2, "Tuple")
insert(3, "Set")
insert(4, "Dictionary")
insert(5, "String")

dict
```

```python
import pandas as pd
dict=pd.DataFrame({"A":[1,2,3,4,5],"B":[6,7,8,9,10]})
dict
```

```python
dict.insert(1,"List",[1,3,5,7,9])
print(dict)
```

```python
dict.insert(3,"Copy_A",dict["A"])
print(dict)
```

```python
dict["Slice_A"]=dict["A"][:3]
print(dict)
```

# `Delete`

- we can delete items from a dictionary using the `pop` method

```python
dict = {"A":1, "B":2, "C":3, "D":4, "E":5}
dict.pop("A")
print(dict)
```

```python
import pandas as pd
list=pd.DataFrame({"A":[1,2,3,4],"B":[5,6,7,8],"C":[10,11,12,13]})
print(list)
```

```python
list.pop("A")
print(list)
```

```python
del list["C"]
list
```
